```python
from langchain_core.prompts  import ChatPromptTemplate
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables      import RunnableLambda, RunnableBranch, RunnablePassthrough

from velari_ai.integrations.langchain.utils import build_chat_model
from velari_ai.ai.types import ModelConfig
```

```python
model_config, model = build_chat_model(model_config=ModelConfig(model="gpt-4o-mini", extra={"temperature": 0.5}))
```

## Prompt Templates

```python
prompt_template = ChatPromptTemplate.from_messages(
    [
        SystemMessage(
            "You are an ethical hacker and exploit developer that creates obfuscation techniques for " \
            "this type of vulnerability {vulnerability_type}.",
        ),
        HumanMessage("Create {obfuscation_count} obfuscation technique examples."),
    ]
)
prompt_value = prompt_template.invoke(
    {"obfuscation_count": 3, "vulnerability_type": "XSS"}
)
display(prompt_value)
display(prompt_value.to_messages())
result: AIMessage = model.invoke(prompt_value)
result.content
```

```python
# from a templated string
template        = "Explain the security concept {topic} on the {vulnerability}."
prompt_template = ChatPromptTemplate.from_template(template)
prompt_value    = prompt_template.invoke({"topic": "XSS attacks", "vulnerability": "web"})
display(prompt_value)

result: AIMessage   = model.invoke(prompt_value)
result.content
```

```python
# Define additional processing steps using RunnableLambda
prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are an ethical hacker providing guidance on {topic}."),
        ("human", "Provide {tip_count} techniques and examples of tools."),
    ]
)

uppercase_output = RunnableLambda(lambda x: str(x).upper())
count_words      = RunnableLambda(lambda x: f"Word count: {len(str(x).split())}\n{str(x)}")
chain            = prompt_template | model | StrOutputParser() | uppercase_output | count_words
result           = chain.invoke({"topic": "scanning a web application", "tip_count": 5})
```
